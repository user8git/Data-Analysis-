```python
# Plotting Top 10 medal winning Countries
Top_Countries = df.value_counts(["Country_Code"]).reset_index(name="Counts")[:10]
print(Top_Countries)
```

```python
sns.set_style('darkgrid')
matplotlib.rcParams['figure.figsize'] = (12, 10)

plt.title("Overall Participation Country-Wise")
barplot = sns.barplot(data=Top_Countries, x="Country_Code", y="Counts", hue="Country_Code", palette="CMRmap", dodge=False)
barplot.set_xticklabels(rotation=50, labels=Top_Countries.Country_Code)
plt.ylabel("Counts")
plt.show()
```

```python
# Gender-distribution amongst winners
Gender_Counts = df.value_counts(["Gender"]).reset_index(name= "Counts")[:10]
print(Gender_Counts)

```

```python
#To get the percentage
Gender_Counts['Percentage %'] = round(Gender_Counts.Counts / Gender_Counts.Counts.sum() * 100, 2)

Gender_Counts
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Get gender percentages
Gender_percentage = Gender_Counts["Percentage %"]
Gender_labels = ["Men", "Women"]

# Get colors automatically from seaborn based on gender categories
palette = sns.color_palette("deep", n_colors=len(Gender_labels))

plt.figure(figsize=(5, 5))

# Create pie chart
plt.pie(
    Gender_percentage,
    labels=Gender_labels,
    autopct="%0.1f%%",
    pctdistance=0.80,
    colors=palette,
    textprops={"fontsize": 14}
)

# Create donut hole
centre_circle = plt.Circle((0, 0), 0.70, color='white')
plt.gca().add_artist(centre_circle)

plt.title("Gender Distribution", fontsize=22)
plt.axis("equal")   # Keeps the donut circular
plt.show()
```

```python
# Calculation for trend w.r.t women participation
female_participants = df[(df["Gender"] == "Women")][["Gender", "Year"]]

#group by year
female_participants = female_participants.groupby('Year').count().reset_index()
female_participants
```

```python
# sns.set_style('darkgrid')
plt.figure(figsize=(10,6))
sns.lineplot(data=female_participants, x='Year', y='Gender', color= "red")
plt.title("Female Participation over the Years ")

plt.show()
```

```python
Women_gold_medalists = df[(df["Medal"] == "Gold") & (df["Gender"] == "Women")][["Sport", "Medal"]]
Women_gold_medalists = Women_gold_medalists.value_counts(["Sport"]).reset_index(name = "Women_gold_medalists")[:5]
Women_gold_medalists
```

```python
plt.title("Women  Gold Medalists in Summer Olympics")
barplot = sns.barplot(data=Women_gold_medalists, x="Sport", y="Women_gold_medalists", hue="Sport", palette="CMRmap", dodge=False)
barplot.set_xticklabels(labels=Women_gold_medalists.Sport)
plt.ylabel("Women_gold_medalists")
plt.show()
```
